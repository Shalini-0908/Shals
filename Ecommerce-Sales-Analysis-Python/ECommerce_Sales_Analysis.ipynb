**Exploratory Data Analysis (EDA)**

**Revenue Trend Analysis**

```python
monthly_revenue = revenue.groupby('month')['revenue_usd'].sum()

monthly_revenue.plot(kind='line', marker='o')

plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (USD)')
plt.grid(True)

plt.show()
```

Monthly Orders Trend

```python
monthly_orders = revenue.groupby('month')['orders'].sum()

monthly_orders.plot(kind='line', marker='o')

plt.title('Monthly Orders Trend')
plt.xlabel('Month')
plt.ylabel('Number of Orders')
plt.grid(True)

plt.show()
```

Top 10 Products by Revenue

```python
top_products = orders.groupby('product_name')['total_amount_usd'].sum()

top_products.sort_values(ascending=False).head(10).plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Top 10 Products by Revenue')
plt.xlabel('Product')
plt.ylabel('Revenue (USD)')
plt.show()
```

Revenue by Category

```python
category_revenue = orders.groupby('category')['total_amount_usd'].sum()

category_revenue.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Revenue by Category')
plt.xlabel('Category')
plt.ylabel('Revenue (USD)')
plt.show()
```

**Customer Analysis**

Customer Churn Distribution

```python
customers['churned'].value_counts().plot(
    kind='bar',
    figsize=(6,4)
)

plt.title('Customer Churn Distribution')
plt.xlabel('Churn Status')
plt.ylabel('Number of Customers')
plt.show()
```

**Membership Tier Analysis**

```python
customers.groupby('membership_tier')[
    'total_spend_usd'
].mean().plot(
    kind='bar',
    figsize=(6,4)
)

plt.title('Average Spend by Membership Tier')
plt.xlabel('Membership Tier')
plt.ylabel('Average Spend (USD)')
plt.show()
```

**Customer Age Distribution**

```python
customers['age'].hist(
    bins=20,
    figsize=(8,4)
)

plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

**Acquisition Channel Analysis**

```python
customers['acquisition_channel'].value_counts().plot(
    kind='bar',
    figsize=(8,4)
)

plt.title('Customer Acquisition Channels')
plt.xlabel('Channel')
plt.ylabel('Customers')
plt.show()
```
